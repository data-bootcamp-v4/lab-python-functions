```python
# 1. Define a function named( initialize_inventory that takes products as a parameter. 
# Inside the function, implement the code for initializing the inventory dictionary using a loop and user input.
def initialize_inventory(products):
    inventory = {}

    for name_of_product in products:
        quantity = int(input(f"Enter the quantity of {name_of_product}: "))
        inventory[name_of_product] = quantity

    return inventory
```

```python
# 2. Define a function named get_customer_orders that takes no parameters. Inside the function, implement the code for prompting the user to enter the product names using a loop. The function should return the customer_orders set


def get_customer_orders():
    customer_orders = set()
    answer = "yes"
    while answer == "yes":
        product = input("add product").lower()
        customer_orders.add(product)
        answer = input("do you want another product yes/no: ")
    print ("thanks for your order")
    
    return customer_orders


```

```python
#3. Define a function named update_inventory that takes customer_orders and inventory as parameters. Inside the function, implement the code for updating the inventory dictionary based on the customer orders.


def update_inventory(customer_orders, inventory):
    for product in customer_orders:
        inventory[product]-=1
    return(inventory)

```

```python
#4.  Define a function named calculate_order_statistics that takes customer_orders and products as parameters. Inside the function, implement the code for calculating the order statistics (total products ordered, and percentage of unique products ordered). The function should return these values
def calculate_order_statistics(customer_orders, products):
    products_orders = len(customer_orders)
    percentage_ordered = (products_orders / len(products)) * 100
    return (products_orders, percentage_ordered)

```

```python
# 5. Define a function named print_order_statistics that takes order_statistics as a parameter. Inside the function, implement the code for printing the order statistics. 
def print_order_statistics(order_statistics):
    for order in order_statistics:
        print(order)
```

```python
# 6. Define a function named print_updated_inventory that takes inventory as a parameter. Inside the function, implement the code for printing the updated inventory.

def print_updated_inventory(inventory):
    for item, quantity in inventory.items():
        print(f"{item}: {quantity}")
```

```python
# 7. Call the functions in the appropriate sequence to execute the program and manage customer orders.
products = ["cup", "mug", "shirt", "cap", "shoes"]
```

```python
inventory = initialize_inventory(products)
customer_orders = get_customer_orders()
inventory = update_inventory(customer_orders, inventory)
order_statistics = calculate_order_statistics(customer_orders, products)
print_order_statistics(order_statistics)
print_updated_inventory(inventory)

```
