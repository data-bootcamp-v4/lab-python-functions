```python
FUNCTION initialize_inventory(products):
    CREATE empty inventory dict
    FOR each product in products:
        LOOP until valid quantity entered
        STORE in inventory
    RETURN inventory

FUNCTION get_customer_orders():
    CREATE empty customer_orders set
    ASK how many products the customer wants to order (validated)
    FOR that many times:
        LOOP until valid product name entered
        ADD to customer_orders
    RETURN customer_orders

FUNCTION update_inventory(customer_orders, inventory):
    FOR each product in customer_orders:
        SUBTRACT 1 from inventory[product]
    (no return needed — dictionary is modified in place)

FUNCTION calculate_order_statistics(customer_orders, products):
    total = number of items in customer_orders
    percentage = (total / number of items in products) * 100
    RETURN (total, percentage)

FUNCTION print_order_statistics(order_statistics):
    UNPACK order_statistics into total and percentage
    PRINT them in the required format

FUNCTION print_updated_inventory(inventory):
    FOR each product, quantity in inventory:
        PRINT product and quantity

MAIN PROGRAM:
    CALL initialize_inventory
    CALL get_customer_orders
    CALL update_inventory
    CALL calculate_order_statistics
    CALL print_order_statistics
    CALL print_updated_inventory
```

```python
#Remember our products are: 
products = ["t-shirt", "mug", "hat", "book", "keychain"]

#Our inventory is:
inventory = {"t-shirt":10, "mug":12, "hat":2, "book":3, "keychain":35}
```

```python
# List of products
products = ["t-shirt", "mug", "hat", "book", "keychain"]


# 1. Function to set up inventory
def initialize_inventory(products):
    inventory = {}
    for product in products:
        quantity = input("Enter quantity of " + product + ": ")
        quantity = int(quantity)
        inventory[product] = quantity
    return inventory


# 2. Function to get customer orders
def get_customer_orders():
    customer_orders = set()
    for i in range(3):
        order = input("Enter a product the customer wants to order: ")
        customer_orders.add(order)
    return customer_orders


# 3. Function to update inventory
def update_inventory(customer_orders, inventory):
    for product in customer_orders:
        inventory[product] = inventory[product] - 1


# 4. Function to calculate statistics
def calculate_order_statistics(customer_orders, products):
    total_products_ordered = len(customer_orders)
    percentage_ordered = (total_products_ordered / len(products)) * 100
    return total_products_ordered, percentage_ordered


# 5. Function to print statistics
def print_order_statistics(order_statistics):
    total_products_ordered = order_statistics[0]
    percentage_ordered = order_statistics[1]
    print("Order Statistics:")
    print("Total Products Ordered:", total_products_ordered)
    print("Percentage of Products Ordered:", percentage_ordered, "%")


# 6. Function to print inventory
def print_updated_inventory(inventory):
    print("Updated Inventory:")
    for product in inventory:
        print(product, ":", inventory[product])


# 7. Run the program
inventory = initialize_inventory(products)
customer_orders = get_customer_orders()

order_statistics = calculate_order_statistics(customer_orders, products)
print_order_statistics(order_statistics)

update_inventory(customer_orders, inventory)
print_updated_inventory(inventory)
```
