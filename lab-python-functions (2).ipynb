```python
# List of available products

products = ["t-shirt", "mug", "hat", "book", "keychain"]
```

```python
#1. Define a function named initialize_inventory that takes products as a parameter

def initialize_inventory(products):
    inventory = {}

    for product in products:
        quantity = int(input(f"Enter the quantity of {product}: "))
        inventory[product] = quantity

    return inventory
```

```python
#2. Define a function named get_customer_orders that takes no parameters.

def get_customer_orders():
    customer_orders = set()

    while True:
        product = input("Enter the product you want to order (or 'done' to finish): ").lower()

        if product == "done":
            break
        elif product in products:
            customer_orders.add(product)
        else:
            print("Invalid product. Please try again.")

    return customer_orders
```

```python
#3. Define a function named update_inventory that takes customer_orders and inventory as parameters. 

def update_inventory(customer_orders, inventory):
    
# Update inventory by reducing quantities based on customer orders.
   
    for product in customer_orders:
        if product in inventory:
            if inventory[product] > 0:
                inventory[product] -= 1
            else:
                print(f"Sorry, {product} is out of stock!")
    
```

```python
#4. Define a function named calculate_order_statistics that takes customer_orders and products as parameters. 


def calculate_order_statistics(customer_orders, products):
    total_products_ordered = len(customer_orders)

    percentage_ordered = ((total_products_ordered / len(products)) * 100 )

    return total_products_ordered, percentage_ordered
```
