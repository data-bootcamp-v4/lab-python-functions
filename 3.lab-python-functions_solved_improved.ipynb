```python
def initialize_inventory (products):
    inventory = {}
    for product in products:
        inventory[product] = int(input(f"How many {product} do we have?"))
    print(f"The inicial inventory is: {inventory}")
    return inventory
```

```python
def get_customer_orders():
    customer_orders = set()
    while True:
        product_name = input("What would you like to order?")
        if product_name not in products:
            answer = input("We don't have this product here, would you like something else? Yes or No?").lower().strip()
            if answer == "no":
                break
        else:
            customer_orders.add(product_name)
            answer = input("Would you like something else? Yes or No?").lower().strip()
            if answer == "no":
                break
    print(f"The customer order is: {customer_orders}")
    return customer_orders
```

```python
def update_inventory(customer_orders,inventory):
    for products, quantity in inventory.items():
        if products in customer_orders:
            inventory[products] = quantity -1
    print(f"The updated inventory is: {inventory}")
    return inventory
```

```python
def calculate_order_statistics(customer_orders, products):
    order_statistics = (len(customer_orders),round(len(customer_orders)/len(products)*100,2))
    return order_statistics
```

```python
def print_order_statistics (order_statistics):
    print ()
    print ("Order Statictics:")
    print(f"Total products ordered: {order_statistics[0]}")
    print(f"Percentage of unique products ordered: {order_statistics[1]}%")
```

```python
def print_updated_inventory (inventory):
    print ()
    print("The updated inventory is:")
    for products, quantity in inventory.items():
        print(f"{products}:{quantity}")
```

```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]
inventory = initialize_inventory (products)
customer_orders = get_customer_orders()
inventory = update_inventory(customer_orders,inventory)
order_statistics = calculate_order_statistics(customer_orders, products)
print_order_statistics (order_statistics)
print_updated_inventory (inventory)
```
