## Managing Customer Orders with Functions

This notebook refactors the previous customer-order program into reusable functions.

The products used in the previous lab are kept here so that the new functions reproduce the same behavior.

```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]
```

### 1. Initialize the inventory

```python
def initialize_inventory(products):
    """Create an inventory dictionary with quantities entered by the user."""
    inventory = {}

    for product in products:
        while True:
            try:
                quantity = int(input(f"Insert quantity of {product}: "))
                if quantity < 0:
                    print("Please enter a non-negative quantity.")
                    continue
                inventory[product] = quantity
                break
            except ValueError:
                print("Please enter a whole number.")

    return inventory
```

### 2. Get customer orders

```python
def get_customer_orders():
    """Collect valid, unique customer orders until the user chooses to stop."""
    customer_orders = set()

    while True:
        order = input("Write the product you want: ").strip()

        if order in products:
            customer_orders.add(order)
        else:
            print("Item not found")

        continue_order = input("Do you want to continue? yes/no: ").strip().lower()
        if continue_order == "no":
            break

    return customer_orders
```

### 3. Update the inventory

```python
def update_inventory(customer_orders, inventory):
    """Reduce the inventory by one unit for every ordered product."""
    for order in customer_orders:
        if order in inventory:
            inventory[order] -= 1

    return inventory
```

### 4. Calculate order statistics

```python
def calculate_order_statistics(customer_orders, products):
    """Calculate the total and percentage of unique products ordered."""
    total_products_ordered = len(customer_orders)

    if len(products) == 0:
        percentage_unique_products_ordered = 0.0
    else:
        percentage_unique_products_ordered = (
            total_products_ordered / len(products) * 100
        )

    return total_products_ordered, percentage_unique_products_ordered
```

### 5. Print the order statistics

```python
def print_order_statistics(order_statistics):
    """Display the order statistics."""
    total_products_ordered, percentage_unique_products_ordered = order_statistics

    print(f"Total products ordered: {total_products_ordered}")
    print(
        f"Percentage of unique products ordered: "
        f"{percentage_unique_products_ordered:.2f}%"
    )
```

### 6. Print the updated inventory

```python
def print_updated_inventory(inventory):
    """Display every product and its remaining quantity."""
    print("Updated inventory:")
    for product, quantity in inventory.items():
        print(f"{product}: {quantity}")
```

## Execute the program

Set RUN_INTERACTIVE to True when you want to enter the quantities and orders yourself. It is False by default so the notebook can be run without stopping for input.

```python
RUN_INTERACTIVE = False

if RUN_INTERACTIVE:
    inventory = initialize_inventory(products)
    customer_orders = get_customer_orders()
    update_inventory(customer_orders, inventory)
    order_statistics = calculate_order_statistics(customer_orders, products)
    print_order_statistics(order_statistics)
    print_updated_inventory(inventory)
```

## Example execution

This example checks the complete sequence without requiring manual input.

```python
example_inventory = {product: 2 for product in products}
example_orders = {"mug", "book", "hat"}

update_inventory(example_orders, example_inventory)
example_statistics = calculate_order_statistics(example_orders, products)
print_order_statistics(example_statistics)
print_updated_inventory(example_inventory)
```
